```python
# Working with Strings
str1 = """this is also a string which you didn't know"""

# We use three different ty pes for different characters to accomodate in that string.


# Escape Characters:
str2 = "This is line 1. \nThis is line 2."
print(str2)

str3 = "This is with \t space."
print(str3)
```

```python
# String Concatenation

print("hello" + "world")

# Length of string
print(len(str1))
```

```python
# Slicing in Strings

str4 = "Divya D. Sharma" 
print(str4[0:5])
print(str4[6:8])
print(str4[9:len(str4)])

# Note: when we don't write a value like [:4] or [4:] then for the first param, it is replaced by 0 and in the second param it is replaced by the last index
```

```python
# Slicing using Negative Index

str5 = "Banana"
# B -> -6
# a -> -5
# ...
# n -> -2
# a -> -1

print(str5[-6: -3])
```

```python
# String Functions:

print(str5.endswith("a"))
print(str5.capitalize())
print(str5.replace("B", "D"))
print(str5.find("a"))
print(str5.count("a"))

# Since all these functions don't change the string.
print(str5)
```
