```python
# Slicing in Strings

str4 = "Divya D. Sharma" 
print(str4[0:5])
print(str4[6:8])
print(str4[9:len(str4)])

# Note: when we don't write a value like [:4] or [4:] then for the first param, it is replaced by 0 and in the second param it is replaced by the last index
```

```python
# Slicing using Negative Index

str5 = "Banana"
# B -> -6
# a -> -5
# ...
# n -> -2
# a -> -1

print(str5[-6: -3])
```

```python
# String Functions:

print(str5.endswith("a"))
print(str5.capitalize())
print(str5.replace("B", "D"))
print(str5.find("a"))
print(str5.count("a"))

# Since all these functions don't change the string.
print(str5)
```

```python
# Dictionaries in Python
```

```python
#  Data is stored in key:value pairs 
#  Unordered, mutable and don't allow duplicates.

exampleDict = {
    "name": "Divya",
    "gpa": 4.0,
    "marks": [75,74,74,74,68,65]
}
print(exampleDict)
print(exampleDict["gpa"])
```

```python
# Modifying a Dictionary
exampleDict["surname"] = "Sharma"
print(exampleDict)


# Nested Dictionary
exampleDict["subjects"] = {
    "phy": 74,
    "chem": 74,
    "math": 75,
    "eng": 68,
    "computer": 74,
    "nepali": 65
}

print(exampleDict)
```

```python
#  Dictionary methods

print(exampleDict.keys()) # return all keys
exampleDict.values() # returns all the values
exampleDict.items() # returns all (key,val) paris as tuples
exampleDict.get("name") # takes the parameter of the name of the key and returns the value
# exampleDict.update("new") # inserts the specified items to the dictionary
```

```python
#  to convert dictionary to tuple or list
# Type casting

print(list(exampleDict.keys()))
```
